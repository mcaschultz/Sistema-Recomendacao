# Lista dos 5 Filmes mais bem avaliados

```python
# Caminho para o arquivo contendo o nome dos filmes.
caminho_filmes = r"dados\dado3.txt"
```

```python
# Lê arquivo de filmes.
with open(caminho_filmes, 'r', encoding='utf-8') as f:
    filmes = f.readlines()
```

```python
# Câlculo das notas médias de cada filme.
notas_medias = np.mean(gradiente_conjugado.notas_preditas(), axis=1)
print(notas_medias)
```

```python
# Calculo dos indices dos 5 maiores filmes.
indices_top5 = np.argpartition(notas_medias, -5)[-5:]
indices_ordem = indices_top5[np.argsort(notas_medias[indices_top5])[::-1]]

print(indices_ordem)
```

```python
# 5 filmes em ordem de maior para o menor.
for idx in indices_ordem:
    print(f"Filme: {filmes[idx]} Nota média: {notas_medias[idx]}")
```
